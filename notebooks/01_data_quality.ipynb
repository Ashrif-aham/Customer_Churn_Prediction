```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
# Notebook is inside the notebooks/ folder
PROJECT_ROOT = Path("..")

DATA_PATH = (PROJECT_ROOT/ "Customer_Churn_Prediction-"/ "WA_Fn-UseC_-Telco-Customer-Churn.csv")

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
OUTPUT_DIR = PROJECT_ROOT / "outputs" / "metrics"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Current working directory:", Path.cwd())
print("Dataset path:", DATA_PATH)
print("Dataset exists:", DATA_PATH.exists())
```

```python
df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())
```

```python
df.info()
```

```python
df.describe(include="all").T
```

```python
missing_values = df.isnull().sum().sort_values(ascending=False)

missing_values
```

```python
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).sort_values(ascending=False)

missing_summary = pd.DataFrame({
    "Missing Count": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary
```

```python
print("Duplicate rows:", df.duplicated().sum())

print(
    "Duplicate customer IDs:",
    df["customerID"].duplicated().sum()
)
```

```python
for column in df.columns:
    if df[column].nunique() <= 20:
        print("\n", column)
        print(df[column].value_counts(dropna=False))
```

```python
object_columns = df.select_dtypes(include="object").columns

for column in object_columns:
    df[column] = df[column].str.strip()
```

```python
print("Before conversion:")
print(df["TotalCharges"].dtype)

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

print("\nAfter conversion:")
print(df["TotalCharges"].dtype)

print(
    "\nMissing TotalCharges:",
    df["TotalCharges"].isna().sum()
)
```

```python
df[df["TotalCharges"].isna()][
    [
        "customerID",
        "tenure",
        "MonthlyCharges",
        "TotalCharges"
    ]
]
```

```python
df[df["TotalCharges"].isna()][
    [
        "customerID",
        "tenure",
        "MonthlyCharges",
        "TotalCharges"
    ]
]
remaining_missing = df["TotalCharges"].isna().sum()

print("Remaining missing values:", remaining_missing)
```

```python
new_customer_mask = (
    df["TotalCharges"].isna()
    & (df["tenure"] == 0)
)

df.loc[
    new_customer_mask,
    "TotalCharges"
] = 0
```

```python
remaining_missing = df["TotalCharges"].isna().sum()

print("Remaining missing values:", remaining_missing)
```

```python
print("Total missing values in dataset:")
print(df.isna().sum())
```

```python
numeric_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

df[numeric_columns].describe()
```

```python
for column in numeric_columns:
    invalid_count = (df[column] < 0).sum()

    print(
        column,
        "- Negative values:",
        invalid_count
    )
```

```python
print(df["Churn"].value_counts(dropna=False))
```

```python
valid_churn_values = ["Yes", "No"]

invalid_churn = df[
    ~df["Churn"].isin(valid_churn_values)
]

print(
    "Invalid churn rows:",
    len(invalid_churn)
)
```

```python
before = len(df)

df = df.drop_duplicates(
    subset="customerID",
    keep="first"
).copy()

after = len(df)

print(
    "Duplicate records removed:",
    before - after
)
```

```python
quality_summary = pd.DataFrame({
    "Metric": [
        "Number of Rows",
        "Number of Columns",
        "Total Missing Values",
        "Duplicate Rows",
        "Duplicate Customer IDs"
    ],
    
    "Value": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["customerID"].duplicated().sum()
    ]
})

quality_summary
```

```python
quality_summary.to_csv(
    OUTPUT_DIR / "data_quality_summary.csv",
    index=False
)

print("Data quality summary saved.")
```

```python
CLEANED_DATA_PATH = (
    PROCESSED_DIR
    / "customer_churn_cleaned.csv"
)

df.to_csv(
    CLEANED_DATA_PATH,
    index=False
)

print(
    "Cleaned dataset saved to:",
    CLEANED_DATA_PATH
)
```
